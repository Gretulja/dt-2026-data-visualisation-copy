Create a chart showing who puts in the most effort (hours of study)

```python
import pandas as pd
df = pd.read_csv("Competencies.csv")
df.head()
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="Student", y="Hours_Studied", legend=False)
avg_hours= df.groupby("Student")["Hours_Studied"].mean() #use this for an average in a bar chart
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied", color="#FFC0CB")
colors= plt.cm.plasma(range(len(avg_hours)))
plt.title("Which student put in the most effort?")
plt.ylabel("Hours studied")
plt.show()
#idk mis siin selle sinisega juhtus aga okei
```
