Here I created a bar chart to visualise the average score of each book. Based on this chart, we can see that most books on this list have the rating of four, but ratings five and three are also present for some books.

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Book title", y="Average_score_on_internet", legend=False)
plt.title("The average score of a book")
plt.ylabel("Average score")
plt.show()
```

Here I made a bubble chart to see if there is any correlation between a book's release year and its score. We can see that there really isn't any correlation, but we can note that all four star books were released in the 1990s or after. We can also see that the two books from before the 1990s have either a high or a low rating.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("5. digitalhumanities dataset.csv")
plt.scatter(df["Year"], df["Average_score_on_internet"], s=df["Average_score_on_internet"]*20, alpha=0.5)
plt.title("Years books were released with their scores")
plt.ylabel("Average score")
plt.xlabel("Release year")
```

I will now use a boxplot to see the distribution of the years the books were released. Based on the boxplot, we can see that the median of the release years is somewhere around 2005. We can also see that the release years have two outliers from before 1970.

```python
import seaborn as sns
sns.boxplot(data=df, x="Year")
plt.title("Distribution of the years the books were released")
plt.show()
```

I will find the average score of each publisher. We can see that the publisher with the highest scores are Scholastic Press and Viking Press. The publishers with the lowest scores are Dutton Children's Books and Little, Brown and Company. The one self-published book also has a low rating. Farrar, Straus and Giroux have an average score of 3.75, and Simon & Schuster have an average score of 4.25. Rest of the publishers have an average rating of four.

```python
avg_score= df.groupby("Publisher")["Average_score_on_internet"].mean()
avg_score.plot(kind="bar")
plt.ylabel("Average score")
plt.title("The average score of each publisher")
plt.show()
```
