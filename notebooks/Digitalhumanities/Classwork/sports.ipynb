```python
# can use to work with tabular data
import pandas as pd
df = pd.read_csv("competition.csv")
df.head() # Show the first few rows
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.title("Years of practice per athlete")
plt.ylabel("Years of practice")
plt.show()
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Age per athlete")
plt.ylabel("Age of practice")
plt.show()
```

```python
df.plot (kind="pie", y="Age",labels=df["Name"], legend=False, autopct="%1.1f%%")
plt.title("Athletes' age")
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*60, alpha=0.6)
# *60 means if a student has more years of practice their bubble is bigger
# alpha=0.6 makes the bubbles a bit transparent
plt.title("Years of practice with students' age")
plt.ylabel("Years of practice")
plt.xlabel("Age")
```

```python
import seaborn as sns #use this to create cleaner and more formatted plots
sns.boxplot(data=df, x="Age")
plt.title("Distribution of athletes' age")
plt.show()
```

```python
plt.Figure(figsize=(8,5))
df[["Age", "Years_of_practice"]].boxplot()
plt.title("Box plot for age and years of practice")
plt.show
```

```python
#group the sport by age
avg_age= df.groupby("Sport")["Age"].mean()
avg_age.plot(kind="bar")
plt.ylabel("Average age")
plt.show()
```

```python
#group the sport by years of practice
avg_age= df.groupby("Sport")["Years_of_practice"].mean()
avg_age.plot(kind="bar")
plt.ylabel("Average years of practice")
plt.show()
```
