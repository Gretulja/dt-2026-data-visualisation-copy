For task 2, I am going to create a wordcloud and describe it afterwards.

```python
# Install required packages
!pip install wordcloud matplotlib

# Import libraries
import os
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from google.colab import files

# Check if file exists, if not, ask to upload
filename = "17.10 homework text file.txt"
if not os.path.exists(filename):
    print(f"Faili '{filename}' ei leitud. Palun laadige see uuesti üles:")
    uploaded = files.upload()

# Read text file
with open(filename, "r", encoding="utf-8") as file:
    text = file.read()

# Generate word cloud
wordcloud = WordCloud(width=600, height=400, background_color="white").generate(text)

# Display the generated image
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```

Based on this wordcloud, we can see that the most common words in the text are PLA and force. A lot of use is also for words technology, experts and rocket. The least used words seem to be new, combat, owned and capabilities. Therefore we can assume that this text talks about PLA and how it's involved with some kind of force or using force, as well as what experts are doing with rocket technology.
