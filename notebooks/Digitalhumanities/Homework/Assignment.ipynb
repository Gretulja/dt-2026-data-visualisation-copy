```python
import pandas as pd
df = pd.read_csv("5. digitalhumanities dataset.csv")
df.head()
```

Here I created a bar chart to visualise the average score of each book. Based on this chart, we can see that most books on that list have the rating of 4, but ratings 5 and 3 are also present for some books.

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Book title", y="Average_score_on_internet", legend=False)
plt.title("The average score of a book")
plt.ylabel("Average score")
plt.show()
```
