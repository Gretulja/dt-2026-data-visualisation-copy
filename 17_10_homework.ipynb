```python
#Import my text file
from wordcloud import WordCloud
import matplotlib.pyplot as plt

with open("17.10 homework text file.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Customise wordcloud
wc = WordCloud(width=600, height=400, background_color="white").generate(text)
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")

plt.show()
```

Based on this wordcloud, we can see that the most common words in the text are PLA and force. A lot of use is also for words technology, experts and rocket. The least used words seem to be new, combat, owned and capabilities. Therefore we can assume that this text talks about or reports things PLA has said and how it's involved with some kind of force or using force. We can also assume it  reports on what either experts have said about rocket technology or how experts are involved with rocket technology.
